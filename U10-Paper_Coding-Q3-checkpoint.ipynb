```python
#For the list1 and list2, use the nested for loop to multiply each element of list 1 and list2
#Print the result with the element multiplication result
#Declare list1 and list2 in the 1st and 2nd rows
#Use the nested for loop in the 3rd and 4th rows
#Use the print loop in the 5th row

list1 = [3,5,7]
list2 = [2,3,4,5,6]
for i in list1:
    for j in list2:
        print(f"{i}*{j}={i*j}")
```

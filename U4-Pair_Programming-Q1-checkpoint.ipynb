```python
#Receive radius value from user
#Print circumference and area of circle with radius value
#Variable PI = 3.141592
#Circumference of circle = 2*radius*PI
#Area of circle = PI*radius*radius

PI = 3.141592
Radius = int(input("Enter the radius of a circle:"))
Circumference = 2*Radius*PI
Area = PI*Radius*Radius

print("Circumference of a circle=",Circumference, ", Area of a circle=", Area)
```

```python
#Write a program that executes addition, subtraction, multiplication and division
#Print the operation result of two positive integers, based on the desired operation number given as an input
#If a number else than 1, 2, 3 and 4 is given as an input, "Entered an incorrect number" is printed
#To enter two numbers, write one, press enter and write another one

print("1) Addition  2) Subtraction  3) Multiplication  4) Division")
operation = int(input("Enter the desired number of operation:"))
if operation >= 1 and operation <=4:
    print("Enter two numbers for operation.")
    integer1 = int(input(""))
    if integer1 >= 0:
        integer2 = int(input(""))
        if integer2 >= 0:
            if operation == 1: #addition
                print(integer1,"+",integer2,"=",integer1+integer2)
            elif operation == 2: #subtraction
                print(integer1,"-",integer2,"=",integer1-integer2)
            elif operation == 3: #multiplication
                print(integer1,"*",integer2,"=",integer1*integer2)
            elif operation == 4: #division
                print(integer1,"/",integer2,"=",integer1/integer2)
        else:
            print("Enter a positive integer.")
    else:
        print("Enter a positive integer.")
else:
    print("Entered an incorrect number.")
```

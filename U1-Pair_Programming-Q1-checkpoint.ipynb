```python
print("   ","*") #3 spaces on each side, 1 asterisk
print("  ","***") #2 spaces on each side, 3 asterisks
print(" ","*****") #1 space on each side, 5 asterisks
print("*********") #0 spaces, 9 asterisks
print(" ","*****") #1 space on each side, 5 asterisks
print("  ","***") #2 spaces on each side, 3 asterisks
print("   ","*") #3 spaces on each side, 1 asterisk
```

```python
print(" "*3,"*") #3 spaces on each side, 1 asterisk
print(" "*2,"*"*3) #2 spaces on each side, 3 asterisks
print(" ","*"*5) #1 space on each side, 5 asterisks
print("*"*9) #0 spaces, 9 asterisks
print(" ","*"*5) #1 space on each side, 5 asterisks
print(" "*2,"*"*3) #2 spaces on each side, 3 asterisks
print(" "*3,"*") #3 spaces on each side, 1 asterisk
```

```python
#Return the element with the maximum number of occurrences
#When there are more than two frequent elements, print the highest number

given_tuples = (1,2,5,4,3,2,1,4,7,8,9,9,3,7,3,9)
counts = {}
for n in given_tuples:
    counts[n] = counts.get(n,0) + 1
frequent = None
max_count = 0
for n, count in counts.items():
    if count > max_count or (count == max_count and n > frequent):
        max_count = count
        frequent = n
print("Given tuples:",given_tuples)
print("The most frequent element:",frequent)
```

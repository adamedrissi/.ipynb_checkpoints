```python
#String s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
#If the string lengths are the same, write a program that prints all of the three shortest strings
#Use the sort(key=len) function to sort the strings by length and then write a code

s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
s_list.sort(key=len)
shortest_len = len(s_list[0])
shortest_strings = [f"'{s}'" for s in s_list if len(s) == shortest_len]
print("The shortest strings:",",".join(shortest_strings))
```

```python
#Pythagorean Theorem: c**2 = a**2 + b**2
#c=hypotenuse, a=height, b=bottom
#Write code that calculates the length of c by receiving a and b as integers
#Enter b in the 1st line and a in the 2nd line
#In the 3rd line, print out the result of calculating the length of c
import math

height = int(input("Height of triangle: "))
bottom = int(input("Bottom of triangle: "))
hypotenuse = math.sqrt(height**2 + bottom**2)

print("Hypotenuse of triangle: ",hypotenuse)
```

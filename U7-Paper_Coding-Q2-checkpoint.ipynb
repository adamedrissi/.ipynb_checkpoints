```python
#Write a program which receives two integers a, b as input
#Determines whether a is a multiple of b
#Print the result

a,b = map(int,input("Write two integers:").split())
if a%b == 0:
    print(a,"is a multiple of",b)
else:
    print(a,"is not a multiple of",b)
```

```python
#Integers entered in separate lines

a = int(input("Write two integers:"))
b = int(input(""))
if a%b == 0:
    print(a,"is a multiple of",b)
else:
    print(a,"is not a multiple of",b)
```

```python
#Write a program that generates a multidimensional array of n x n size, based on the number of inputs, receiving two or more n as inputs from users
#The content arrangement should be displayed so that the values of 0 and 1 intersect in a checkered pattern

n = int(input("Enter n:"))
if n<2:
    print("Please enter a number larger than or equal to 2.")
else:
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    for i in range(n):
        for j in range(n):
            if (i+j) % 2 == 0:
                matrix[i][j] = 1
            else:
                matrix[i][j] = 0
    for i in matrix:
        print(" ".join(map(str,i)))
```

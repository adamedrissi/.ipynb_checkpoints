```python
#Receive a 3-digit integer from user
#If the first digit of the integer n is 3, return True
#If the first digit of the integer n is not 3, return False
#// operator must be familiar

n = int(input("Enter an integer:"))
if n>=100 and n<=999:
    print(n//100 == 3)
else:
    print("The integer must be a 3-digit number.")
```

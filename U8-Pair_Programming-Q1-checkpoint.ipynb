```python
#Agency A is planning to issue tickets of a concert hall for idol singers' concert
#The number n is the input
#n*n seats are placed when n is given as input
#The arrangement of seat numbers is called a snake matrix
#The snake matrix increases the array in ones in the shape of a snake's trunk

n = int(input("Enter n:"))
step = 1
for row in range(n):
    if row % 2 == 0:
        for col in range(n):
            print(f"{step + col:2}",end=" ")
    else:
        for col in range(n):
            print(f"{step + n-1 - col:2}",end=" ")
    step += n
    print()
```

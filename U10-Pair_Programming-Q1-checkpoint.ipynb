```python
#String s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
#Do not use the min function or sort method to print the shortest string
#Print the string that shows the first if there are multiple shortest strings

s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
shortest = s_list[0]
for s in s_list:
    if len(s) < len(shortest):
        shortest = s
print("The shortest string:",shortest)
```

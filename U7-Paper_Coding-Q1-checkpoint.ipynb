```python
#Receive an alphabetical letter from the user
#Print "It is a vowel" for a,e,i,o,u
#Print "It is a consonant" for any other letters

letter = str(input("Enter the alphabet:"))
if letter == "a" or letter == "e" or letter == "i" or letter == "o" or letter == "u":
    print("It is a vowel")
else:
    print("It is a consonant")
```

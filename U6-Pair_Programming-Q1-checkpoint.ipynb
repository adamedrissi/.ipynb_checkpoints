```python
#Write a program that receives age as input
#Print "Adult if age is 20 or over
#Print "Youth" if age is under 20 and equal to or over 10
#Print "Kid" if age is under 10

age = int(input("Enter age:"))
if age >= 20:
    print("Adult")
elif age < 20 and age >= 10:
    print("Youth")
else:
    print("Kid")
```

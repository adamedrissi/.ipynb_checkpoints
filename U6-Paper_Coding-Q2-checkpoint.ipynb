```python
#Write a program that receives any integer x between -100 and 100
#1) Prints x on the screen
#2) Prints "...is a natural number" if x is an integer greater than zero
#Otherwise, let it simply print x

x = int(input("Enter integer:"))
if x>=-100 and x<=100:
    print("x =",x)
    if x>0:
        print(x,"is a natural number.")
else:
    print("The integer must be between -100 and 100.")
```

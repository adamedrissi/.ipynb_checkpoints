```python
#Develop a menu ordering program for Yummy Restaurant
#Menu is shown to user
#Let the user select an item
#If given input alphabet is not in menu, print "Enter the menu again" and receive another input
#Menu items: Burger(b), Chicken(c), Pizza(p)

print("Welcome to Yummy Restaurant. Here is the menu.")
print("- Burger(enter b)")
print("- Chicken(enter c)")
print("- Pizza(enter p)")
while True:
    menu = str(input("Choose a menu item (enter b, c or p):"))
    if menu == "b": #burger
        print("You chose burger.")
        break
    elif menu == "c": #chicken
        print("You chose chicken.")
        break
    elif menu == "p": #pizza
        print("You chose pizza.")
        break
    else: #wrong input
        print("Enter the menu item again.")
```

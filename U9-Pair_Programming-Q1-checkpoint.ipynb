```python
#Palindrome number refers to an integer whose value is the same as its original value, even if listed upside down
#e.g. 121 or 3443
#Write the program to determine whether the number is a palindrome or not by receiving the number n from the user

n = int(input("Enter an integer:"))
if n < 0:
    print(f"{n} is not a palindrome.") #A negative number is not a palindrome
else:
    string_n = str(n)
    if string_n == string_n[::-1]:
        print(f"{n} is a palindrome.")
    else:
        print(f"{n} is not a palindrome.")
```

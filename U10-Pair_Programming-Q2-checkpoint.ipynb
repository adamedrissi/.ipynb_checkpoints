```python
#String s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
#Do not use the max function or sort method to print the longest string
#Print the string that shows the first if there are multiple longest strings

s_list = ['abc','bcd','bcdefg','abba','cddc','opq']
longest = s_list[0]
for s in s_list:
    if len(s) > len(longest):
        longest = s
print("The longest string:",longest)
```

```python
#Write a program that receives a point with x and y coordinates as input
#Determines in which quadrant among 1,2,3,4 the point belongs
#positive y and negative x is 2 quadrant
#positive y and positive x is 1 quadrant
#negative y and negative x is 3 quadrant
#negative y and positive x is 4 quadrant
#neutral y and/or neutral x is no applicable quadrant

x,y = map(int,input("Enter x,y coordinates:").split())
if y > 0 and x < 0: #2 quadrant
    print("In the second quadrant")
elif y > 0 and x > 0: #1 quadrant
    print("In the first quadrant")
elif y < 0 and x < 0: #3 quadrant
    print("In the third quadrant")
elif y < 0 and x > 0: #4 quadrant
    print("In the fourth quadrant")
else: #neutral
    print("No applicable quadrant")
```

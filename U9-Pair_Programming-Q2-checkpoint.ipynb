```python
#The computer has a random integer between 1 and 100 as the correct answer value
#When the user presents an appropiate answer, the program only informs whether the presented integer is higher or lower compared the correct random answer value
#The game is repeated until the user guesses the answer correctly

import random

correct_answer = random.randint(1,100)
tries = 0
print("Guess a number between 1 to 100")
while True:
    user_answer = int(input("Enter a number:"))
    if user_answer == correct_answer: #correct entry between 1 to 100
        tries += 1
        print("Congratulations. Total tries =",tries)
        break
    elif user_answer < 1 or user_answer > 100:
        print("Enter an appropriate answer. A number between 1 to 100.")
    else:
        tries += 1
        if user_answer > correct_answer:
            print("Lower!")
        else:
            print("Higher!")
```

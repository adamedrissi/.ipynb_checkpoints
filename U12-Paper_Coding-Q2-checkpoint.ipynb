```python
#The tuple records daily sales of a store for 10 days
#Write a code to print how many days had reduced sales compared to the previous day
#Compare the values by iterating the elements with the iteration statement

daily_sales_record = (100, 121, 120, 130, 140, 120, 122, 123, 190, 125)
reduced_sales_days = 0
for i in range(1, len(daily_sales_record)):
    if daily_sales_record[i] < daily_sales_record[i-1]:
        reduced_sales_days += 1
print("Daily sales record:",daily_sales_record)
print(f"In the past {len(daily_sales_record)} days, {reduced_sales_days} days had reduced sales compared to the previous day.")
```
